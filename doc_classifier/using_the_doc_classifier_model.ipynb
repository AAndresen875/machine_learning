```python
#imports 
import os
import pickle
import pandas as pd,
from sklearn.datasets import load_files
```

# Loading in the ML model

```python
#loading in ml model trained in the doc classifier creation file
with open('file_classifier', 'rb') as classifying_model:
    model = pickle.load(classifying_model)
model
```

# Using the model to characterize files

```python
def sort_new_files():
    #pointing to the correct directory
    new_data_path = os.path.join(os.getcwd(), 'new_files')
    print(new_data_path)
    
    #using sklearn's load_files which needs the files to be located in a named folder
    new_data_to_sort = load_files(new_data_path, encoding='cp1252')
    print("Amount of new files to sort:", new_data_to_sort.data)
    
    #using the loaded model to classify the files in the unsorted folder:
    newly_classified = model.predict(new_data_to_sort.data)
    
    #making a dataframe with new information:
    results_df = pd.DataFrame(new_data_to_sort.filenames, columns=['file_names', 'file_type'])
    results_df['file_type']=newly_classified
    return results_df

new_classifying = sort_new_files()
new_classifying.head()
```
